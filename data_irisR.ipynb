```r
# 1. Instalar y cargar las librerías nativas de R
if (!require("rpart")) install.packages("rpart")
if (!require("rpart.plot")) install.packages("rpart.plot")
library(rpart)
library(rpart.plot)

# 2. Cargar los datos
data(iris)

# 3. Entrenar el modelo
modelo_arbol <- rpart(Species ~ ., data = iris, method = "class",
                      control = rpart.control(minsplit = 2, cp = 0.01))

# 4. Mostrar la estructura jerárquica en texto
print(modelo_arbol)

# 5. Dibujar el árbol de manera visual
rpart.plot(modelo_arbol, type = 2, extra = 101, fallen.leaves = TRUE, main = "Árbol de Decisión Iris")
```
