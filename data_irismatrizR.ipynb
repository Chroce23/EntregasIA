```r
# 1. Asegurar la reproducibilidad (para que las 20 observaciones al azar sean fijas)
set.seed(42)

# 2. Cargar librerías necesarias
if(!require(e1071)) install.packages("e1071")
library(e1071)

# 3. Cargar el conjunto de datos Iris y entrenar Naive Bayes
data(iris)
modelo_nb <- naiveBayes(Species ~ ., data = iris)

# 4. Seleccionar aleatoriamente 20 observaciones (índices del 1 al 150)
indices_muestra <- sample(1:150, 20)
muestra_evaluacion <- iris[indices_muestra, ]

# 5. Probar el clasificador NB sobre estas 20 observaciones
predicciones_muestra <- predict(modelo_nb, muestra_evaluacion)
valores_reales <- muestra_evaluacion$Species

# 6. Definir los niveles con el orden específico solicitado por el profesor:
# Filas y columnas: (setosa, virginica, versicolor)
orden_profesor <- c("setosa", "virginica", "versicolor")

predicciones_factor <- factor(predicciones_muestra, levels = orden_profesor)
reales_factor <- factor(valores_reales, levels = orden_profesor)

# 7. Construir la Matriz de Confusión (Filas = Predicho, Columnas = Real)
# Tal como pide el ejemplo: si dio setosa y el real es versicolor -> fila setosa, col versicolor
matriz_confusion <- table(Predicho = predicciones_factor, Real = reales_factor)

# 8. Mostrar los resultados en pantalla
print("=== LAS 20 OBSERVACIONES ELEGIDAS ALEATORIAMENTE ===")
print(muestra_evaluacion[, c("Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width", "Species")])

print("=== MATRIZ DE CONFUSIÓN DE 3x3 SOLICITADA ===")
print(matriz_confusion)
```
